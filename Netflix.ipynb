```python
df2=df[['release_year',"description"]]
```

```python
df2
```

```python
for index,row in df2.iterrows():
    d=row['description']
    testimonial=TextBlob(d)
    p=testimonial.sentiment.polarity
    if p==0:
        sent="Neutral"
    elif p>0:
        sent="Positive"
    else:
        sent="Negative"
    df2.loc[[index,2],'Sentiment']=sent
    
```

```python
df2
```

```python
df2=df2[df2['release_year']>=2000]
```

```python
df2
```

```python
sentiment_count=df2.groupby(['release_year','Sentiment']).size().reset_index(name="Total Count")
```

```python
sentiment_count
```

```python
barchart=px.bar(sentiment_count, x="release_year",y="Total Count",color="Sentiment", title='Sentimental analysis of Netflix shows')
```

```python
barchart.show()
```
